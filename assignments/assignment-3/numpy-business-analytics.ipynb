## 1. Creating Arrays

#### By: William Orazi
#### Description:

* Create a 1D array of 100 random integers between 1 and 500.
* Reshape the 1D array into a 10x10 matrix.

```python
# Generates an array of random integers with a given count from given
# low to high values and optional seed.
# 
# Returns a 1D array of integers
def generate_random_integers(count, low, high, seed=1):
    random_number_generator = np.random.default_rng(seed)

    return random_number_generator.integers(low, high + 1, size=count)

# Create 1D array of 100 integers between 1 and 500
random_values = generate_random_integers(100, 1, 500, RANDOM_SEED)

print("1D Array:\n")
print(random_values);

# Reshape array into a 10x10 matrix
matrix = random_values.reshape(10, 10)

print("\n10x10 Matrix:\n")
print(matrix)
```
