# Part 1: Data Manipulation and Analysis

## 1. Creating Arrays (William Orazi)

* Create a 1D array of 100 random integers between 1 and 500.
* Reshape the 1D array into a 10x10 matrix.

```python
# Generates an array of random integers with a given count from given
# low to high values and optional seed.
# 
# Returns a 1D array of integers
def generate_random_integers(count, low, high, seed=1):
    if count <= 0:
        print("Error: Count must be greater than 0.")
        return None

    if low > high:
        print("Error: Low cannot be greater than high.")
        return None
    
    random_number_generator = np.random.default_rng(seed)

    return random_number_generator.integers(low, high + 1, size=count)
```

```python
# Create 1D array of 100 integers between 1 and 500
random_values = generate_random_integers(100, 1, 500, RANDOM_SEED)

print("1D Array:\n")
print(random_values);

if random_values is not None:
    # Reshape array into a 10x10 matrix
    matrix = random_values.reshape(10, 10)

    print("\n10x10 Matrix:\n")
    print(matrix)
```

## 2. Array Slicing and Indexing

* Extract all elements from the third column of the matrix.
* Extract the submatrix consisting of rows 2 through 5 and columns 4 through 8.

```python
# Extracting the third column
third_column = matrix[:, 2]

print("3rd Column:\n")
print(third_column)

# Rows 2-5 map to indices 1-4, and columns 4-8 map to indices 3-7
submatrix = matrix[1:5, 3:8]

print("\nSubmatrix (Rows 2-5, Columns 4-8):\n")
print(submatrix)
```

## 3. Filtering Data

* Find all elements in the matrix that are greater than 200.
* Replace all elements in the matrix that are less than 100 with a value of 0.

```python
# Find all elements greater than 200
greater_than_200 = matrix[matrix > 200]

print("Elements > 200:\n")
print(greater_than_200)

copied_matrix = matrix.copy()

# Mask to find all elements less than 100
less_than_100 = copied_matrix < 100

# Replace elements less than 100 with 0
copied_matrix[less_than_100] = 0

print("\nMatrix with values < 100 replaced by 0:\n")
print(copied_matrix)
```

## 4. Basic Statistical Analysis

* Calculate the mean, median, standard deviation, and variance of the entire matrix.
* Calculate the mean of each row and each column.

```python
print("Mean:", np.mean(matrix))
print("Median:", np.median(matrix))
print("Standard Deviation:", np.std(matrix))
print("Variance:", np.var(matrix))

print()

print("Row Means:", np.mean(matrix, axis=1))
print("Column Means:", np.mean(matrix, axis=0))
```

## 5. Sorting

* Sort each row of the matrix in ascending order.
* Sort the entire matrix by the values in the second column.

```python
# Sorting across rows in ascending order
matrix_with_sorted_rows = np.sort(matrix, axis=1)

print("Rows sorted in ASC order:\n")
print(matrix_with_sorted_rows)

# Sort the entire matrix by the values in the second column
matrix_sorted_by_second_column = matrix[np.argsort(matrix[:, 1])]

print("\nMatrix sorted by the second column:\n")
print(matrix_sorted_by_second_column)
```
